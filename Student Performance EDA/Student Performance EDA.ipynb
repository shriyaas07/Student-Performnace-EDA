```python
sns.countplot(x='sex', data=df)
plt.title('Gender Distribution')
plt.show()
```

Insight:
There are more female students than male students in the dataset.


```python
sns.histplot(df['age'], kde=True, bins=8)
plt.title('Age Distribution')
plt.show()
```

Insight:
Most students are aged between 15 and 18, with a slight skew toward younger ages.


```python
sns.boxplot(x='studytime', y='G3', data=df)
plt.title('Study Time vs Final Grade (G3)')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')
plt.show()
```

Insight:
Students who spend more time studying tend to perform better in their final grade.


```python
sns.barplot(x='internet', y='G3', data=df)
plt.title('Internet Access vs Final Grade (G3)')
plt.xlabel('Internet Access')
plt.ylabel('Final Grade (G3)')
plt.show()
```

Insight:
Students with internet access appear to have higher final grades on average.


```python
sns.regplot(x='G1', y='G3', data=df)
plt.title('First Period Grade (G1) vs Final Grade (G3) with Trendline')
plt.xlabel('G1')
plt.ylabel('G3')
plt.show()
```

Insights:
The scatter plot with a fitted regression line clearly indicates a strong positive linear relationship between G1 and the G3.
As G1 increases, G3 also tends to increase. This suggests that a student's performance early in the term can be a strong predictor of their final performance.

```python
plt.figure(figsize=(12, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

Insinghts:
Final grade (G3) is strongly correlated with G1 (0.83) and G2 (0.92), positively influenced by study time (0.25), and negatively impacted by past failures (-0.39), indicating that early performance, consistent study habits, and academic history play a key role in final outcomes.
